```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("../data/raw/monthly_updates.csv")
df.head()
```

```python
df.info()
```

```python
df.columns = ["month", "total_updates", "cumulative_updates"]
df.head()
```

```python
# Convert month column safely
df["month"] = pd.to_datetime(df["month"], errors="coerce")

# Remove invalid rows (if any)
df = df.dropna(subset=["month"])

# Sort by time
df = df.sort_values("month").reset_index(drop=True)

print(df)
```

```python
df.describe()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["month"], df["total_updates"], marker="o")
plt.title("Monthly Aadhaar Updates Trend")
plt.xlabel("Month")
plt.ylabel("Total Updates")
plt.grid(True)
plt.show()
```
