```python
import pandas as pd

df = pd.read_csv("../data/raw/monthly_updates.csv")
df.columns = ["month", "total_updates", "cumulative_updates"]

# Safe datetime conversion
df["month"] = pd.to_datetime(df["month"], errors="coerce")

# Remove invalid rows
df = df.dropna(subset=["month"])

# Sort chronologically
df = df.sort_values("month").reset_index(drop=True)

df.head()



```

```python
df["rolling_avg"] = df["total_updates"].rolling(window=3).mean()
```

```python
df["trend_deviation"] = df["total_updates"] - df["rolling_avg"]
```

```python
df.fillna(0, inplace=True)
df.head()
```

```python
df.to_csv("../data/processed/final_features.csv", index=False)
```
